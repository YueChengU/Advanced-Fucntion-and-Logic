```python
# Problem 1
""" Write a python program that, given an input list of any level of complexity/nestedness, will return the inner most list plus 1. 
This is to be done with a while loop. Note: the input will contain only integers or lists."""

input_list = [1,2,3,4,[5,6,7,[8,9]]]#This is the test list input

# To find the most inner list, I'm using any() to see if there's at least one list nested within the input list, and using isinstance() to check if each item is a list.
# To fetch the most inner list, I'll append all the items in the identified lists into a new list flat_list, which will take them to an outer level

#Since we need to check the data type of the items one by one, I think it would be best to integrated for-loop with while-loop to complete the task.
def innermost_plusone(input_list):
    while any(isinstance(i, list) for i in input_list):#if there's at least one item in the input_list that is a list, isinstance(i, list) will be Rrue and any() condition will remain True
        flat_list = []#flat_list is used as a intermediate list when we break the list up to find the most inner list, which will be clear out for each loop.
        #This loop will run until there's nothing but integers in a list. Then we exit the loop with the most inner list found.
        for i in input_list:#Since we need to check the data type of the items one by one, I think it would be best to integrated for-loop with while-loop to complete the task.
            if isinstance(i, list):#Select only the lists not integers within the original list
                flat_list.extend(i)#Join all the inner lists into one new list thus they are flatten out for one level.
        input_list = flat_list #put all items from the inner level back into the loop.

    return [x + 1 for x in input_list]

innermost_plusone(input_list)       
```
