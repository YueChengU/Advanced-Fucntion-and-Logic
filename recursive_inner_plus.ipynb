```python
# Problem 2

"""Write the a python program that, given an input list of any level of complexity/nestedness, 
will return the inner most list plus 1. This is to be done with recursion. Note: the input will contain only integers or lists. """

input_list = [1,2,3,4,[5,6,7,[8,9]]] #This is the testing list
#The recusive function will basically do the same thing like the while loop function, but by putting the function in the returning result the function will keep running until there's no list within the final list.

def recursive_inner_plus(input_list):
    
    if all(isinstance(i, int) for i in input_list):#When there's no more list and only integers in the list, it has reach the deepest level and is then ready to return the results.
        return [i + 1 for i in input_list]# Plus 1 for the elements in the list

    else:
        flat_list = []
        for i in input_list:
            if isinstance(i, list):#Pick out all the list nested within the input list, and by extend.() merge them all so the become the next level list
                flat_list.extend(i)
                
        return recursive_inner_plus(flat_list)#To return a function result it has to calculate the recursive_inner_plus(flat_list) thus send the flaten list into second round

recursive_inner_plus(input_list)
```
