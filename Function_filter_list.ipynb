```python
# Write a python program that, given an input list, will filter the input above a user defined threshold. This is to be done with a standard function.
input_list = [1,2,3,4,5,6,7,8,9]#This is the testing list
def filter_above(input_list, n):#input_list as the list to be filtered, n as the threshold
    filtered_list = []#Create an empty list to keep the elements that doesn't get filtered  
    #I think the pop.() method to take the number off the original list will make it more complicated with indexing, so here I'm just taking out the elements we want to keep.
    for i in input_list:
        if i <= n:
            filtered_list.append(i) #If the number doesn't exceed the threshold, it will be put in the given list.
        #If the number is larger the the threshold n, it won't be kept and no action is needed.
    return filtered_list

filter_above(input_list, 6)
       
```
